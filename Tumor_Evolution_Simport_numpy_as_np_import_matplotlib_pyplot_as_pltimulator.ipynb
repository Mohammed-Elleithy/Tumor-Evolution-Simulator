# Tumor Evolution Simulator

A computational simulation demonstrating how treatment-resistant cancer clones emerge through evolutionary selection and drive tumor relapse.

---

## Project Overview

Cancer is not a static disease. Tumors consist of heterogeneous populations of cells that continuously evolve under selective pressures such as chemotherapy, immune attack, nutrient limitation, and hypoxia.

This project explores a simplified evolutionary model of cancer progression by simulating two populations of tumor cells:

1. Treatment-sensitive cancer cells
2. Treatment-resistant cancer cells

The simulation demonstrates how therapy can successfully eliminate most sensitive cells while unintentionally selecting for resistant populations that later dominate the tumor and cause relapse.

---

## Scientific Background

One of the central ideas in modern oncology is that cancer behaves as an evolving ecosystem.

Tumors are composed of multiple cellular clones that differ genetically and phenotypically.

When treatment is applied:

* Sensitive cells are eliminated.
* Resistant cells survive.
* Resistant cells expand and become dominant.

This process resembles Darwinian natural selection.

The treatment itself acts as a selection pressure.

---

## Biological Terminology

### Clone

A clone is a population of cells derived from a single ancestral cell.

If one cancer cell acquires a mutation and proliferates, all descendant cells belong to the same clone.

### Tumor Heterogeneity

Tumors are not composed of identical cells.

Different clones may coexist within the same tumor.

Some clones may be sensitive to therapy while others are resistant.

### Selection Pressure

Any factor that causes certain cells to survive better than others.

Examples include:

* Chemotherapy
* Immunotherapy
* Hypoxia
* Nutrient deprivation
* Immune surveillance

### Resistance

The ability of cancer cells to survive treatments that kill other tumor cells.

### Relapse

The return of cancer after an apparent response to treatment.

---

## Research Question

How can a small population of resistant cancer cells survive treatment and eventually drive tumor relapse?

---

## Model Assumptions

### Sensitive Cells

Sensitive cells are highly susceptible to treatment.

Each treatment cycle removes 95% of sensitive cells.

### Resistant Cells

Resistant cells are largely unaffected by treatment.

Only 5% die during each treatment cycle.

### Resistant Cell Growth

Surviving resistant cells continue to proliferate and expand.

### Closed System

The model does not include:

* Immune cells
* Mutations
* Metastasis
* Angiogenesis

These features can be added in future versions.

---

## Expected Biological Outcome

At the beginning of treatment:

* Tumor size decreases rapidly.
* Sensitive cells are eliminated.

However:

* Resistant cells survive.
* Resistant cells expand.
* Resistant cells eventually dominate the tumor population.

This produces relapse.

## Disclaimer

This simulation is an educational model and is not intended to represent the full biological complexity of real tumors.


```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================
# Initial Tumor Population
# ==========================

sensitive = 990
resistant = 10

# ==========================
# Lists to store results
# ==========================

sensitive_history = []
resistant_history = []
total_history = []

# ==========================
# Tumor Evolution Simulation
# ==========================

for cycle in range(50):

    # Treatment kills 95% of sensitive cells
    sensitive = sensitive * 0.05

    # Resistant cells mostly survive treatment
    resistant = resistant * 0.95

    # Resistant cells continue dividing
    resistant = resistant * 1.30

    # Store results for plotting
    sensitive_history.append(sensitive)
    resistant_history.append(resistant)
    total_history.append(sensitive + resistant)

# ==========================
# Plot Results
# ==========================

plt.figure(figsize=(10,6))

plt.plot(total_history, label="Total Tumor Cells")
plt.plot(sensitive_history, label="Sensitive Cells")
plt.plot(resistant_history, label="Resistant Cells")

plt.xlabel("Treatment Cycle")
plt.ylabel("Number of Cells")
plt.title("Tumor Evolution Under Treatment")

plt.legend()
plt.grid(True)

plt.show()
```
